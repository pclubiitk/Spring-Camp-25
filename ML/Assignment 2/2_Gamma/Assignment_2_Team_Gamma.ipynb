```python

def plot_columns(df, col1_index, col2_index):

  col1_name = df.columns[col1_index]
  col2_name = df.columns[col2_index]

  plt.figure(figsize=(10, 6))
  plt.scatter(df.iloc[:, col1_index], df.iloc[:, col2_index])
  plt.xlabel(col1_name)
  plt.ylabel(col2_name)
  plt.title(f'Scatter Plot of {col1_name} vs {col2_name}')
  plt.grid(True)
  plt.show()
#for i in range(0,5):
#  for j in range(0,5):
#   plot_columns(df,i,j)
plot_columns(df,0,4)    #hints towards some inverse relation
plot_columns(df,1,4)
plot_columns(df,2,4)    #hints towards similar type of relation between (E & B) & (E & Q)


```

```python
# prompt: create a fuction which takes index of 2 columns as input and prints the graph of columns 1 vs log of other

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np



def plot_columns_log(df, col1_index, col2_index):
  col1_name = df.columns[col1_index]
  col2_name = df.columns[col2_index]

  # Check if the column for log transformation has non-positive values
  if (df.iloc[:, col2_index] <= 0).any():
      print(f"Warning: Column '{col2_name}' contains non-positive values. Cannot compute logarithm.")
      return

  plt.figure(figsize=(10, 6))
  plt.scatter(df.iloc[:, col1_index], np.log(df.iloc[:, col2_index]))
  plt.xlabel(col1_name)
  plt.ylabel(f'Log({col2_name})')
  plt.title(f'Scatter Plot of {col1_name} vs Log({col2_name})')
  plt.grid(True)
  plt.show()

for i in range(0,5):
  for j in range(0,5):
    print("#####################################################################")
    plot_columns_log(df,i,j)

```

```python

def plot_log_columns(df, col1_index, col2_index):

  col1_name = df.columns[col1_index]
  col2_name = df.columns[col2_index]

  if (df.iloc[:, col1_index] <= 0).any() or (df.iloc[:, col2_index] <= 0).any():
      print(f"Warning: Column '{col1_name}' or '{col2_name}' contains non-positive values. Cannot compute logarithm.")
      return

  plt.figure(figsize=(10, 6))
  plt.scatter(np.log(df.iloc[:, col1_index]), np.log(df.iloc[:, col2_index]))
  plt.xlabel(f'Log({col1_name})')
  plt.ylabel(f'Log({col2_name})')
  plt.title(f'Scatter Plot of Log({col1_name}) vs Log({col2_name})')
  plt.grid(True)
  plt.show()

for i in range(0,5):
  for j in range(0,5):
    plot_log_columns(df,i,j)
```

```python


def plot_column_vs_square(df, col1_index, col2_index):

  col1_name = df.columns[col1_index]
  col2_name = df.columns[col2_index]

  plt.figure(figsize=(10, 6))
  plt.scatter(df.iloc[:, col1_index], df.iloc[:, col2_index]**2)
  plt.xlabel(col1_name)
  plt.ylabel(f'{col2_name}^2')
  plt.title(f'Scatter Plot of {col1_name} vs {col2_name}^2')
  plt.grid(True)
  plt.show()

for i in range(0,5):
  for j in range(0,5):
    plot_column_vs_square(df,i,j)
```

```python


# Create the Ed^2 column
df['Ed^2'] = df['E'] * (df['d']**2)

# Function to plot Ed^2 vs a specified column
def plot_ed2_vs_column(df, col_index):
  col_name = df.columns[col_index]
  plt.figure(figsize=(10, 6))
  plt.scatter(df.iloc[:, col_index], df['Ed^2'])
  plt.xlabel(col_name)
  plt.ylabel('Ed^2')
  plt.title(f'Scatter Plot of Ed^2 vs {col_name}')
  plt.grid(True)
  plt.show()

# Function to plot Ed^2 vs sin of a specified column
def plot_ed2_vs_sin_column(df, col_index):
  col_name = df.columns[col_index]
  plt.figure(figsize=(10, 6))
  plt.scatter(np.sin(df.iloc[:, col_index]), df['Ed^2'])
  plt.xlabel(f'sin({col_name})')
  plt.ylabel('Ed^2')
  plt.title(f'Scatter Plot of Ed^2 vs sin({col_name})')
  plt.grid(True)
  plt.show()

# Function to plot log(Ed^2) vs log of a specified column
def plot_log_ed2_vs_log_column(df, col_index):
  col_name = df.columns[col_index]
  # Check if the column for log transformation has non-positive values
  if (df.iloc[:, col_index] <= 0).any() or (df['Ed^2'] <=0).any():
    print(f"Warning: Column '{col_name}' or 'Ed^2' contains non-positive values. Cannot compute logarithm.")
    return
  plt.figure(figsize=(10, 6))
  plt.scatter(np.log(df.iloc[:, col_index]), np.log(df['Ed^2']))
  plt.xlabel(f'log({col_name})')
  plt.ylabel('log(Ed^2)')
  plt.title(f'Scatter Plot of log(Ed^2) vs log({col_name})')
  plt.grid(True)
  plt.show()

# Function to plot Ed^2 vs square of a specified column
def plot_ed2_vs_square_column(df, col_index):
  col_name = df.columns[col_index]
  plt.figure(figsize=(10, 6))
  plt.scatter(df.iloc[:, col_index]**2, df['Ed^2'])
  plt.xlabel(f'{col_name}^2')
  plt.ylabel('Ed^2')
  plt.title(f'Scatter Plot of Ed^2 vs {col_name}^2')
  plt.grid(True)
  plt.show()

# Example usage: plot Ed^2 vs each of the first 5 columns
for i in range(5):
  plot_ed2_vs_column(df, i)
  plot_ed2_vs_sin_column(df,i)
  plot_log_ed2_vs_log_column(df,i)
  plot_ed2_vs_square_column(df,i)
```
